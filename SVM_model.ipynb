```python
import pandas as pd
import numpy as np
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('/content/waste management.csv')
data.head()
```

```python
x=data[['waste_type','material_composition','recycling_potential','toxicity_level']]
y=data['management_strategy']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=47)
```

```python
model=SVC(kernel='rbf',C=1.0,gamma='scale')
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy:",accuracy)
```

```python
conf_matrix=confusion_matrix(y_test,y_pred)
sns.heatmap(conf_matrix,annot=True,fmt='d',cmap='Blues',xticklabels=['Non-Recyclable','Rcycleable'],yticklabels=['Non-Recycleable','Recycleable'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('confusion')
plt.show()
```
